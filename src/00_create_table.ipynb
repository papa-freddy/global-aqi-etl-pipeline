```python
import pandas as pd

# --- CONFIGURATION ---
# Replace the following with your own source paths
RAW_CITIES_PATH = "/data/raw/worldcities.csv"
RAW_CONTINENTS_PATH = "/data/raw/country_continents.csv"
PROCESSED_OUTPUT_PATH = "/data/processed/capital_city_locations.csv"

city = pd.read_csv(RAW_CITIES_PATH)
cont = pd.read_csv(RAW_CONTINENTS_PATH)
```

```python
city.head()
```

```python
# Filter out city table to only include primary (capital) cities
city_capital = city[city['capital'] == 'primary']
city_capital.head()
```

```python
cont.head()
```

```python
# Join capital city table with continent-country table on country name
final = pd.merge(city_capital, cont, left_on='country', right_on='Country', how='inner')
```

```python
final.head()
```

```python
# Remove extra country column
final_new = final.drop(columns='Country')
final_new.head()
```

```python
final_new.to_csv(PROCESSED_OUTPUT_PATH)
```
